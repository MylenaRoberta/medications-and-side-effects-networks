# Imports

```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
import nx_parallel as nxp
import itertools
```

# Bipartite Network 

```python
bipartite_df = pd.read_csv('../data/processed/bipartite-network-edges.csv')
bipartite_graph = nx.from_pandas_edgelist(bipartite_df, 'source', 'target')

# Open Node properties DataFrame
node_properties_df = pd.read_csv('../data/processed/bipartite-network-nodes.csv')

node_properties_df.head()

# Add node attributes to the bipartite graph
node_type = node_properties_df.set_index("id")["type"].to_dict()

nx.set_node_attributes(bipartite_graph, node_type, "type")
```

```python
# Analyze the bipartite graph
print(f"Number of nodes: {bipartite_graph.number_of_nodes()}")
print(f"Number of edges: {bipartite_graph.number_of_edges()}")

# Connected components
connected_components = list(nx.connected_components(bipartite_graph))
print(f"Number of connected components: {len(connected_components)}")

# Degree distribution
degrees = sorted((d for n, d in bipartite_graph.degree()), reverse=True)
degree_series = pd.Series(degrees)
print("Degree distribution statistics:")
print(degree_series.describe())

print(f"Top 5 nodes by degree: {degrees[:5]}")

print(f"Top 5 nodes with lowest degree: {degrees[-5:]}")
```

```python
CC_sizes = [len(c) for c in connected_components]
CC_sizes_series = pd.Series(CC_sizes)
print("Connected component size statistics:")
print(CC_sizes_series.describe())

# Rank CC by size the largest to smallest
CC_sizes_sorted = sorted(CC_sizes, reverse=True)
print(f"Sizes of the connected components: {CC_sizes_sorted}")
```

```python
# Show the number of edges in each connected component from largest to smallest
for i, component in enumerate(sorted(connected_components, key=len, reverse=True)):
    subgraph = bipartite_graph.subgraph(component)
    max_degree = max(d for n, d in subgraph.degree())
    print(f"Connected Component {i}: Size = {len(component)}, Edges = {subgraph.number_of_edges()}, Max degree = {max_degree}")
```

```python
# Plot the power law of the number of nodes vs degree distribution in log-log scale
plt.figure(figsize=(8, 6))
cumulative_sum = np.cumsum(degree_series)

plt.plot(degree_series, range(1, len(degree_series) + 1), marker="o")
plt.xscale('log')
plt.yscale('log')
plt.xlabel("Degree")
plt.ylabel("Number of nodes")
plt.title("Degree distribution")
plt.grid(True)
plt.show()
```

```python
avg_clustering = nx.bipartite.average_clustering(bipartite_graph)
print(f"Average clustering coefficient : {avg_clustering}")
print(f"Diameter of the largest connected component: {nx.diameter(largest_cc)}")
```

```python
top_nodes, bottom_nodes = nx.bipartite.sets(largest_cc)
len(top_nodes)
```

```python
betweenness_centraty = nx.bipartite.betweenness_centrality(largest_cc, nodes=bottom_nodes)
closeness_centraty = nx.bipartite.closeness_centrality(largest_cc, nodes=bottom_nodes)
degree_centraty = nx.bipartite.degree_centrality(largest_cc, nodes=bottom_nodes)
```

```python
# create a dataframe to store the centrality measures for nodes in the largest connected component
centrality_df = pd.DataFrame({
    'node': list(largest_cc.nodes()),
    'betweenness_centrality': [betweenness_centraty[node] for node in largest_cc.nodes()],
    'closeness_centrality': [closeness_centraty[node] for node in largest_cc.nodes()],
    'degree_centrality': [degree_centraty[node] for node in largest_cc.nodes()]
})
centrality_df
```

```python
# save the centrality dataframe to a csv file
centrality_df.to_csv('../data/processed/bipartite-largest-cc-centrality-measures.csv', index=False)
```

```python
# sort the dataframe by betweenness centrality and show top 5 nodes
centrality_df.sort_values(by='betweenness_centrality', ascending=False).head()

# sort the dataframe by closeness centrality and show top 5 nodes
centrality_df.sort_values(by='closeness_centrality', ascending=False).head()

# sort the dataframe by degree centrality and show top 5 nodes
centrality_df.sort_values(by='degree_centrality', ascending=False).head()
```

## Communities

```python
# https://networkx.org/documentation/stable/auto_examples/drawing/plot_clusters.html

communities = nx.community.greedy_modularity_communities(largest_cc)

# communities = nx.community.louvain_communities(largest_cc)

# Compute positions for the node clusters as if they were themselves nodes in a
# supergraph using a larger scale factor
supergraph = nx.cycle_graph(len(communities))
superpos = nx.spring_layout(supergraph, scale=2, seed=429)

# Use the "supernode" positions as the center of each node cluster
centers = list(superpos.values())
pos = {}
for center, comm in zip(centers, communities):
    pos.update(nx.spring_layout(nx.subgraph(largest_cc, comm), center=center, seed=1430))

plt.figure(figsize=(12, 12))

tab_10_colors = plt.get_cmap("tab10").colors

# Nodes colored by cluster
for nodes, clr in zip(communities, tab_10_colors):
    nx.draw_networkx_nodes(largest_cc, pos=pos, nodelist=nodes, node_color=clr, node_size=40)

nx.draw_networkx_edges(largest_cc, pos=pos, width=0.05)

# Add labels
# nx.draw_networkx_labels(largest_cc, pos=pos, font_size=6)

# from adjustText import adjust_text
# texts = [plt.text(x, y, s, fontsize=6) for s, (x, y) in pos.items()]
# adjust_text(texts)


plt.tight_layout()
plt.show()
```
