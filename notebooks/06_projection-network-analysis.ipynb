# **Network and Projections Analysis**

# Imports

```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
import nx_parallel as nxp
import itertools
```

# Side Effects Projection

```python
side_effects_df = pd.read_csv('../data/processed/side-effect-projection-edges.csv')

side_effects_df.drop_duplicates(inplace=True)
side_effects_graph = nx.from_pandas_edgelist(side_effects_df, 'source', 'target')
side_effects_df
```

## Degree and Plots

```python
# Analyze the bipartite graph
print(f"Number of nodes: {side_effects_graph.number_of_nodes()}")
print(f"Number of edges: {side_effects_graph.number_of_edges()}")

# Connected components
connected_components = list(nx.connected_components(side_effects_graph))
print(f"Number of connected components: {len(connected_components)}")

# Degree distribution
degrees = sorted((d for n, d in side_effects_graph.degree()), reverse=True)
degree_series = pd.Series(degrees)
print("Degree distribution statistics:")
print(degree_series.describe())

print(f"Top 5 nodes by degree: {degrees[:5]}")

print(f"Top 5 nodes with lowest degree: {degrees[-5:]}")
```

```python
# Plot the power law of the number of nodes vs degree distribution in log-log scale
plt.figure(figsize=(8, 6))
cumulative_sum = np.cumsum(degree_series)

plt.plot(degree_series, range(1, len(degree_series) + 1), marker="o")
plt.xscale('log')
plt.yscale('log')
plt.xlabel("Degree")
plt.ylabel("Number of nodes")
plt.title("Degree distribution")
plt.grid(True)
plt.show()
```

```python
options = {
    'node_color': 'black',
    'node_size': 1,
    'width': 1,
}

plt.Figure(figsize=(10, 10))


nx.draw(side_effects_graph, with_labels=False, **options)
```

## Centrality

```python
# Clustering coefficient
print(f"Average clustering coefficient: {nx.average_clustering(side_effects_graph)}")
print(f"Diameter of the largest connected component: {nx.diameter(side_effects_graph)}")

# # Betweenness centrality
# betweenness = nx.betweenness_centrality(side_effects_graph)
# top_betweenness = sorted(betweenness.items(), key=lambda x: x[1], reverse=True)[:10]
# print("Top 10 nodes by betweenness centrality:")
# for node, centrality in top_betweenness:    
#     print(f"Node: {node}, Betweenness Centrality: {centrality:.4f}")
```

```python
betweenness = nx.betweenness_centrality(side_effects_graph)
closeness = nx.closeness_centrality(side_effects_graph)
degree = nx.degree_centrality(side_effects_graph)
eigenvector = nx.eigenvector_centrality(side_effects_graph)
percolation = nx.percolation_centrality(side_effects_graph)
```

```python
# create a dataframe to store the centrality measures for nodes in the largest connected component
centrality_df = pd.DataFrame({
    'node': list(side_effects_graph.nodes()),
    'betweenness_centrality': [betweenness[node] for node in side_effects_graph.nodes()],
    'closeness_centrality': [closeness[node] for node in side_effects_graph.nodes()],
    'degree_centrality': [degree[node] for node in side_effects_graph.nodes()],
    'eigenvector_centrality': [eigenvector[node] for node in side_effects_graph.nodes()]
})
centrality_df
```

```python
# sort the dataframe by betweenness centrality and show top 5 nodes
centrality_df.sort_values(by='betweenness_centrality', ascending=False).head()

# sort the dataframe by closeness centrality and show top 5 nodes
centrality_df.sort_values(by='closeness_centrality', ascending=False).head()

# sort the dataframe by degree centrality and show top 5 nodes
centrality_df.sort_values(by='degree_centrality', ascending=False).head()

# sort the dataframe by eigenvector centrality and show top 5 nodes
centrality_df.sort_values(by='eigenvector_centrality', ascending=False).head()
```

## Communities

```python

import community as community_louvain

communities = community_louvain.best_partition(side_effects_graph)

# color the nodes according to their partition
cmap = plt.get_cmap('viridis', max(communities.values()) + 1)
shapes = 'so^>v<dph8'

plt.figure(figsize=(12,8))

# draw the graph
pos = nx.spring_layout(side_effects_graph)

# pos = nx.nx_agraph.graphviz_layout(side_effects_graph, prog="neato")

nx.draw_networkx_edges(side_effects_graph, pos, alpha=0.5)
for node, color in communities.items():
    nx.draw_networkx_nodes(side_effects_graph, pos, [node], node_size=100,
                           node_color=[cmap.colors[color]],
                           node_shape=shapes[color])
```

```python
# https://networkx.org/documentation/stable/auto_examples/drawing/plot_clusters.html

communities = nx.community.greedy_modularity_communities(side_effects_graph)

# communities = nx.community.louvain_communities(side_effects_graph)

# Compute positions for the node clusters as if they were themselves nodes in a
# supergraph using a larger scale factor
supergraph = nx.cycle_graph(len(communities))
superpos = nx.spring_layout(supergraph, scale=2, seed=429)

# Use the "supernode" positions as the center of each node cluster
centers = list(superpos.values())
pos = {}
for center, comm in zip(centers, communities):
    pos.update(nx.spring_layout(nx.subgraph(side_effects_graph, comm), center=center, seed=1430))

plt.figure(figsize=(12, 12))

# Nodes colored by cluster
for nodes, clr in zip(communities, ("tab:blue", "tab:orange", "tab:green", "tab:red", "tab:purple", "tab:brown", "tab:pink")):
    nx.draw_networkx_nodes(side_effects_graph, pos=pos, nodelist=nodes, node_color=clr, node_size=40)

nx.draw_networkx_edges(side_effects_graph, pos=pos, width=0.05)

# Add labels
# nx.draw_networkx_labels(side_effects_graph, pos=pos, font_size=6)

from adjustText import adjust_text
texts = [plt.text(x, y, s, fontsize=6) for s, (x, y) in pos.items()]
adjust_text(texts)


plt.tight_layout()
plt.show()
```
